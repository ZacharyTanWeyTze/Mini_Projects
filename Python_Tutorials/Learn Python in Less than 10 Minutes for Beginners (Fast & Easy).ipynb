```python
# Math calculations in Python with variables
a = 10
b = 5

addition = a + b
difference = a - b
multiply = a * b
divide = a/b
power = a**b

print(addition) 
print(difference)
print(multiply)
print(divide)
print(power)
```

```python
# Logic statements
age = 28
isHappy = True

if age > 21:
    print('You are old!')
elif age == 18:
    print('You are getting old!')
else:
    print('You are still young!')
    
if isHappy:
    print('You are happy!')
else:
    print('You are not happy!')
```

```python
#For loop
for i in range(3): # Loop over 3 values
    print('Hello', i) # Print "Hello" and the value of i. 
    
for i in range(3): # Note. In Python programming, index always starts at 0
    print('Hello', i + 1) # Manually add 1 to i to start at 1
```

```python
# Print range object
print(range(3)) # Iterates 3 times
```

```python
# For loops work for lists as well
name_list = ['Luigi', 'Mario', 'Toad']

for name in name_list:
    print(name)
```

```python
# While loop
i = 0

while i < 5:
    i = i + 1 # Increment the value of i by 1 each time it goes in the loop
    print(i) # Print value of i
```

```python
# Create a while True loop (infinite loop)
while True:
    user_input = input('Enter something >> ')
    if user_input == '0':
        print('We are done here.')
        break # Cancel the loop
```

```python
# Function to say hello to the user 
# Uses one function multiple times instead of writing the statement multiple times
def say_hello(name):
    print('Hey there', name)
    
say_hello('Mario')
say_hello('Luigi')
```

```python
# Try Except Block
number = input('Please provide number >> ') # Create an input

try:
    print(10 + int(number)) # Try to print a value plus integer of the number that the user inputs
except:
    print('That is not a valid number!') # Print "not a valid number" if it doesn't work
```
